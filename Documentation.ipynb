**Ma'lumot turlari (Data Type)**

* **VARCHAR**

  *   Limit bo'lganda ishlariladi: VARCHAR (10), VARCHAR(20)
  *  Masalan: 'kitob', 'maktab' vahkz

* **TEXT**
  * Limit bo'lmaganda ishlatiladi

* **CHAR**
  * VARCHAR bilan birxil
  * Farqi har doim berilgan uzunlikda joy oladi. Agar yetishmasa "space" qo'shib qo'yadi

* **INTEGER**
  * **INT**: O'rtacha kattalikdagi butun sonlarni saqlash uchun ishlatiladi. Diapazoni -2,147,483,648 dan 2,147,483,647 gacha.
  * **SMALLINT**: Kichikroq butun sonlar uchun moʻljallangan boʻlib, xotirani tejaydi. Diapazoni -32,768 dan 32,767 gacha.
  * **BIGINT**: Juda katta butun sonlarni (masalan, yirik ID raqamlari yoki hisob-kitoblar) saqlash uchun ishlatiladi. Diapazoni -9,223,372,036,854,775,808 dan 9,223,372,036,854,775,807 gacha.

* **SERIAL** bu oddiy INT (butun son) turi bilan bir xil diapazonga ega, ammo jadvalga yangi qator qo'shilganda qiymatlarni avtomatik ravishda 1 taga oshirib (auto-increment) boradigan maxsus ma'lumot turidir. U odatda jadvallarning birlamchi kaliti (Primary Key) uchun ishlatiladi.
  * **SMALLSERIAL**: Kichik hajmdagi ketma-ketliklar uchun (1 dan 32,767 gacha).
  * **SERIAL**: Standart ketma-ketliklar uchun (1 dan 2,147,483,647 gacha).
  * **BIGSERIAL**: Juda yirik va cheksiz hisoblanadigan jadvallar uchun (1 dan 9,223,372,036,854,775,807 gacha).

* **NUMERIC**
  * **numeric(5, 2)**: Bu yerda birinchi raqam (5) — sonning umumiy raqamlari soni (precision), ikkinchi raqam (2) esa verguldan keyingi raqamlar sonini (scale) bildiradi. Ya'ni, jami 5 ta raqamdan iborat bo'lib, uning 2 tasi kasr qismiga ajratiladi (masalan: 123.45). Agar 123.246 kiritilsa, u verguldan keyin 2 ta raqam qoldirish uchun 123.25 ko'rinishida yaxlitlanadi.

  * **numeric(5)**: Faqat bitta raqam ko'rsatilsa, bu sonning faqat butun qismdan iborat bo'lishini va verguldan keyin raqam bo'lmasligini anglatadi (masalan: 12345). Jami raqamlar soni 5 tadan oshmasligi kerak.

  * **numeric:** Hech qanday parametrlar (argumentlar) berilmasa, ma'lumotlar bazasi cheklovsiz aniqlikdagi har qanday uzunlikdagi butun va o'nlik sonlarni qabul qilaveradi.

* **FLOATING NUMBERS**
  * **REAL (4 bytes)**: Xotiradan 4 bayt joy oladi. U unchalik katta bo'lmagan kasr sonlar va taxminan 6-7 xonali aniqlikdagi raqamlar uchun ishlatiladi.

  * **DOUBLE PRECISION** (8 bytes): Xotiradan 2 barobar ko'p (8 bayt) joy egallaydi. Shuning hisobiga u juda katta kasr sonlarni va yuqoriroq, ya'ni taxminan 15 xonali aniqlikdagi raqamlarni saqlay oladi.

* **BOOLEAN** TRUE/FALSE ma'lumot turi saqlanadi

* **VAQT** ma'lumot turlari **bold text**
  * **DATE** sanalar uchun 29.09.2026
  * **TIME** - vaqt uchun 12:45:40
  * **TIMESTAMP** - sana va vaqt uchun 29.09.2026 12:45:08


# Taqqoslash operatorlari

1. ` SELECT * FROM doughnut_list WHERE price !=1.00;` 1 ga teng bo'lmagan qiymatlarni chiqarish
2. ` SELECT * FROM doughnut_list WHERE price > 3;` 3 dan katta qiymatlarni chiqarish
3. ` SELECT * FROM doughnut_list WHERE price = 3;`3 ga teng qiymatlarni chiqarish

Operatorlar (taqqoslash uchun):
`>` katta, `<` kichik, `>=` katta yoki teng, `<=` kichik yoki teng, `=` teng, `<>` teng emas.

# Buyruqlarni jamlash



**AND va OR** — `WHERE` ichida bir nechta shartni birlashtirish uchun ishlatiladi.

**AND:** hamma shartlar to'g'ri bo'lishi kerak.
**OR:** kamida bitta shart to'g'ri bo'lsa yetadi.

**Misol** (`doughnut_list` jadvali bilan):

| name | price | type |
|---|---|---|
| Glazed | 2.5 | ring |
| Chocolate | 3 | ring |
| Jelly | 4 | filled |
| Cream | 5 | filled |

**AND misoli:**
```sql
SELECT * FROM doughnut_list WHERE price >= 3 AND type = 'filled';
```
Natija: `Jelly`, `Cream`. Ikkala shart ham bajarilishi kerak (narx 3+ **va** turi filled). `Chocolate` narxi mos, lekin turi `ring`, shuning uchun tushib qoladi.

**OR misoli:**
```sql
SELECT * FROM doughnut_list WHERE price < 3 OR type = 'filled';
```
Natija: `Glazed`, `Jelly`, `Cream`. Bitta shart bajarilsa yetadi. `Chocolate` ikkala shartga ham mos kelmaydi, shuning uchun u yo'q.

**Eslab qolish:**
- `AND` = qat'iy (natija **kamroq** bo'ladi)
- `OR` = yumshoq (natija **ko'proq** bo'ladi)

**Muhim tuzoq:** `AND` va `OR` birga ishlatilsa, `AND` birinchi hisoblanadi. Tartibni boshqarish uchun qavs `( )` ishlating:

```sql
WHERE (price < 3 OR price > 4) AND type = 'ring'
```

Qavs bo'lmasa, natija boshqacha chiqishi mumkin. Bu juda ko'p uchraydigan xato.

**Pandas bilan solishtirish:** SQL'dagi `AND` o'rniga `&`, `OR` o'rniga `|` ishlatiladi. Har bir shart alohida qavs ichida bo'lishi shart: `df[(df["price"] >= 3) & (df["type"] == "filled")]`

# LIKE, BETWEEN, IN

1. LIKE: matn bo'yicha qidirish (pattern)

# SQL: `LIKE`, `BETWEEN`, `IN`

## LIKE: matn andozasi
- `%` = istalgan miqdordagi belgi
- `_` = aynan bitta belgi

```sql
WHERE first_name LIKE 'J%'         -- J bilan boshlanadi
WHERE last_name LIKE '%ov'         -- ov bilan tugaydi
WHERE interests LIKE '%football%'  -- ichida football bor
```

Katta-kichik harfga sezgir. Befarq qidirish uchun `ILIKE`.

## BETWEEN: oraliq
```sql
WHERE birthdate BETWEEN '1980-01-01' AND '1990-12-31'
```

Ikkala chegara ham kiradi. Kichik qiymat birinchi yoziladi.

## IN: ro'yxatdan biri
```sql
WHERE profession IN ('doctor', 'driver')
```

Bir nechta `OR` o'rniga ishlatiladi.

## Muhim
- Uchalasiga `NOT` qo'shsa, teskarisi bo'ladi: `NOT LIKE`, `NOT BETWEEN`, `NOT IN`.
- Hech biri `NULL` ni topmaydi. Buning uchun `IS NULL` ishlatiladi

# UPDATE

**Umumiy shakl:**



*   `UPDATE` jadval_nomi
*   `SET` ustun = yangi_qiymat
* `WHERE` shart;

`UPDATE` jadvaldagi **mavjud qatorlarning** qiymatini o'zgartiradi.

**Umumiy shakl:**

```sql
UPDATE jadval_nomi
SET ustun = yangi_qiymat
WHERE shart;
```

**Misollar:**

```sql
-- bitta ustun
UPDATE my_contacts
SET profession = 'teacher'
WHERE first_name = 'John';

-- bir nechta ustun (vergul bilan)
UPDATE my_contacts
SET profession = 'doctor', interests = 'football'
WHERE last_name = 'King';

-- hisoblash bilan (narxni 10% oshirish)
UPDATE doughnut_list
SET price = price * 1.1
WHERE price < 2;
```

**Eng muhim qoida: `WHERE` ni unutma.** `WHERE` yozmasang, **jadvaldagi hamma qator** o'zgaradi. Shuning uchun xavfsiz usul:

1. Avval `SELECT * FROM jadval WHERE shart;` bilan qaysi qatorlar tanlanishini ko'r.
2. Natija to'g'ri bo'lsa, `SELECT *` ni `UPDATE ... SET ...` ga almashtir.

**Bilib qo'yish kerak:**
- `NULL` ni to'ldirish uchun ham `UPDATE` ishlatiladi: `SET birthdate = '1990-05-05' WHERE ...`. Qatorni topish uchun `IS NULL` ham ishlaydi.
- Qiymatni `NULL` qilish: `SET ustun = NULL`.
- `WHERE` ichida oldingi mavzularni (`LIKE`, `IN`, `BETWEEN`) ishlatsa bo'ladi.
- PostgreSQL da `RETURNING *` oxiriga qo'shilsa, o'zgargan qatorlar darhol ko'rsatiladi.
- Natija: `UPDATE 2` degani 2 ta qator o'zgargan.


# DELETE

`DELETE` jadvaldan **qatorlarni** o'chiradi. Ustunni emas, jadvalning o'zini ham emas.

**Umumiy shakl:**

```sql
DELETE FROM jadval_nomi
WHERE shart;
```

**Misollar:**

```sql
-- bitta qatorni o'chirish
DELETE FROM my_contacts
WHERE first_name = 'Tom';

-- shart bilan (tug'ilgan sanasi yo'qlar)
DELETE FROM my_contacts
WHERE birthdate IS NULL;

-- oldingi mavzular bilan
DELETE FROM my_contacts
WHERE profession IN ('driver', 'doctor');
```

**Eng muhim qoida: `WHERE` ni unutma.** `WHERE` yozmasang, **jadvaldagi hamma qator** o'chadi (jadval qoladi, lekin bo'sh bo'ladi).

**Xavfsiz usul** (`UPDATE` dagi kabi):

1. Avval `SELECT * FROM jadval WHERE shart;` bilan qaysi qatorlar tanlanishini ko'r.
2. Natija to'g'ri bo'lsa, `SELECT *` ni `DELETE` ga almashtir.

**Bilib qo'yish kerak:**
- `DELETE` o'chirilgan qatorlarni **qaytarib bo'lmaydi** (pgAdmin da avtomatik commit bo'ladi).
- Natija: `DELETE 2` degani 2 ta qator o'chgan.
- Oxiriga `RETURNING *` qo'shsang, o'chirilgan qatorlar ko'rsatiladi.
- Farqi:

| Buyruq | Nimani o'chiradi |
|---|---|
| `DELETE FROM t WHERE ...` | tanlangan qatorlarni |
| `DELETE FROM t` | hamma qatorni (jadval qoladi) |
| `TRUNCATE t` | hamma qatorni (tezroq) |
| `DROP TABLE t` | jadvalning o'zini ham |

# PRIMARY KEY

`PRIMARY KEY` (asosiy kalit) jadvaldagi **har bir qatorni noyob tarzda aniqlaydigan** ustun.

**Qoidalari:**
1. **Takrorlanmaydi** (`UNIQUE`). Ikkita qatorda bir xil qiymat bo'lishi mumkin emas.
2. **`NULL` bo'lmaydi** (`NOT NULL`).
3. Jadvalda **bitta** primary key bo'ladi (lekin u bir nechta ustundan iborat bo'lishi mumkin).

```sql
CREATE TABLE doughnut_list (
  doughnut_id SERIAL PRIMARY KEY,
  doughnut_name VARCHAR(50),
  doughnut_type VARCHAR(50),
  price NUMERIC(4,2)
);
```

`SERIAL` = raqam avtomatik 1, 2, 3... bo'lib oshib boradi. Qo'lda yozish shart emas.

Bu faqat ustunda takroriy yoki `NULL` qiymat bo'lmasa ishlaydi.

**Bilib qo'yish kerak:**
- Primary key ga takroriy qiymat qo'shsang: `duplicate key value violates unique constraint` xatosi chiqadi.
- Primary key ni olib tashlash: `ALTER TABLE t DROP CONSTRAINT constraint_nomi;`
- Keyinchalik jadvallarni bog'lashda (`JOIN`, `FOREIGN KEY`) primary key juda kerak bo'ladi.
